# Задание
В урне 5 белых и 4 черных шара. Из урны наугад вынимают два шара. Какова вероятность того, что это будут: а) два белых шара; б) два черных шара; в) один черный и один белый.

В данной задаче можно использовать комбинаторику для нахождения вероятностей вытаскивания шаров определенного цвета.

Шаг 1: Определение общего числа возможных исходов
Всего в урне 5 белых и 4 черных шара. Общее количество шаров в урне равно 5 + 4 = 9

Мы выбираем два шара из этих девяти. Число всех возможных C(9, 2) = 36

Шаг 2: Вычисление вероятностей
а) Вероятность вытащить два белых шара
Число способов выбрать 2 белых шара из 5 равно C(5, 2) = 10

Таким образом, вероятность вытащить два белых шара равна отношению C(9, 2) к C(5, 2) = 5/18

б) Вероятность вытащить два черных шара
Число способов выбрать 2 черных шара из 4 равна C(4, 2) = 6

Вероятность вытащить два черных шара равна отношению C(9, 2) к C(4, 2) = 1/6

в) Вероятность вытащить один белый и один черный шар
Число способов выбрать 1 белый шар из 5 и 1 черный из 4 равна произведению числа сочетаний C(5, 1) * C(4, 1) = 20

Вероятность вытащить один белый и один черный шар равна отношению полученного числа к полной вероятности: 20/36

```python
import math
import matplotlib.pyplot as plt

# Общее число способов выбрать 2 шара из 9
total_combinations = math.comb(9, 2)

# а) Вероятность вытащить два белых шара
white_combinations = math.comb(5, 2)
prob_two_white = white_combinations / total_combinations

# б) Вероятность вытащить два черных шара
black_combinations = math.comb(4, 2)
prob_two_black = black_combinations / total_combinations

# в) Вероятность вытащить один белый и один черный шар
one_white_one_black_combinations = math.comb(5, 1) * math.comb(4, 1)
prob_one_white_one_black = one_white_one_black_combinations / total_combinations

# Визуализация графика с полученными вероятностями
probabilities = [prob_two_white, prob_two_black, prob_one_white_one_black]
labels = ['Два белых шара', 'Два черных шара', 'Один белый и один черный шар']

# Построение графика с кривыми
plt.figure(figsize=(8, 6))

# Данные для кривых
x = [1, 2, 3]  # Позиции для каждой вероятности на оси X
y = probabilities  # Вероятности для оси Y

# Построение графика кривой
plt.plot(x, y, marker='o', linestyle='-', color='b', label='Вероятности', linewidth=2, markersize=8)

# Настройки осей и графика
plt.xticks(x, labels)  # Заменяем метки на оси X
plt.title('Вероятности вытаскивания шаров разного цвета из урны')
plt.ylabel('Вероятность')
plt.ylim(0, 1)
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Показать значения вероятностей на графике
for i, prob in enumerate(probabilities):
    plt.text(x[i], prob + 0.02, f'{prob:.2f}', ha='center', fontsize=12)

# Показать график
plt.show()
```
